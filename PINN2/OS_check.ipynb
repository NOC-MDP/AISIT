```python
import icechunk as ic
import xarray as xr
```

```python
def readonly_session(bucket,prefix,snapshot_id=None,history=False,branch="main"):

    storage = ic.s3_storage(
        bucket=bucket,
        prefix=prefix,
        endpoint_url="https://noc-msm-o.s3-ext.jc.rl.ac.uk",
        anonymous=True,
        force_path_style=True, 
    )
    
    repo_config2 = ic.RepositoryConfig(
    storage = ic.StorageSettings(
        unsafe_use_conditional_update=False,
        unsafe_use_conditional_create=False,
    )
    )
    
    repo = ic.Repository.open(storage=storage,config=repo_config2)
    if history:
        print([ancestor for ancestor in repo.ancestry(branch=branch)])
    if snapshot_id:
        return repo.readonly_session(snapshot_id=snapshot_id)
    else:
        return repo.readonly_session(branch=branch)
```

```python
session = readonly_session(bucket="conservative-tracers",prefix="Ba",history=True,snapshot_id="8FMVMWB0VC2K9R97FGT0")

ds = xr.open_zarr(
    session.store,
    consolidated=False
)

print(ds)
```
